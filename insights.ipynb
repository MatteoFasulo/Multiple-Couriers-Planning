```python
import os, json, re
import pandas as pd
```

```python
folder = 'res/CP/'
files = os.listdir(folder)

df = pd.DataFrame(columns=['solver', 'instance', 'time', 'optimal', 'obj'])

for file in files:
    data = json.load(open(folder + file))
    gecode_data = data['gecode']
    time = gecode_data['time']
    optimal = gecode_data['optimal']
    obj = gecode_data['obj']
    df.loc[len(df)] = ['gecode', int(re.search(r'\d+', file).group()), time, optimal, obj]

df.sort_values(by=['instance'], inplace=True)
```

```python
df.tail(7)
```

```python
# Plot gecode objective values and times

df.plot(x='instance', y='obj', kind='bar', title='Gecode objective values')
```

```python
folder = 'res/CP/'
files = os.listdir(folder)

df = pd.DataFrame(columns=['solver', 'instance', 'time', 'optimal', 'obj'])

for file in files:
    data = json.load(open(folder + file))
    try:
        gecode_data = data['chuffed']
    except KeyError:
        continue
    time = gecode_data['time']
    optimal = gecode_data['optimal']
    obj = gecode_data['obj']
    df.loc[len(df)] = ['gecode', int(re.search(r'\d+', file).group()), time, optimal, obj]

df.sort_values(by=['instance'], inplace=True)
```

```python
# Plot gecode objective values and times

df.plot(x='instance', y='obj', kind='bar', title='Chuffed objective values')
```
